```python
def soma(saldo):
  num1 = float(input("Digite o primeiro valor da soma: "))
  num2 = float(input("Digite o segundo valor da soma: "))
  final = num1 + num2
  return print(f"A soma dos números {num1} + {num2} e igual a {final}")

def sub(saldo):
  num1 = float(input("Digite o valor da soma: "))
  num2 = float(input("Digite o valor da soma: "))
  final = num1 - num2
  return print(f"A subtração dos números {num1} + {num2} e igual a {final}")

def div(saldo):
  num1 = float(input("Digite o valor da soma: "))
  num2 = float(input("Digite o valor da soma: "))
  if num2 != 0:
    final = num1 / num2
    return print(f"A divisão dos números {num1} + {num2} e igual a {final}")
  else:
    return print("Não e possivel dividir por zero amigão")

def mult(saldo):
  num1 = float(input("Digite o valor da soma: "))
  num2 = float(input("Digite o valor da soma: "))
  final = num1 * num2
  return print(f"A multiplicação dos números {num1} + {num2} e igual a {final}")
```
